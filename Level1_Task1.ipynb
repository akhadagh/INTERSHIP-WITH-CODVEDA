```python
#Starting with data cleaning and preprocessing on (iris.csv dataset) for level 1 Task 1

#Importing necessary libraries
import pandas as pd

#Loading the dataset
df = pd.read_csv(r'/home/akhada/Downloads/Data Set For Task/1) iris.csv')


```

```python
#Creating Visualizations for EDA

#Importing necessary libraries for visualization (Histogram)
import matplotlib.pyplot as plt

df["sepal_length"].hist()
plt.title("Distribution of Sepal Length in Histogram")
plt.xlabel("Sepal_Length")
plt.ylabel("Frequency")
plt.show()
```

```python
#Importing necessary libraries for visualization (Histogram)

import seaborn as sns

sns.boxplot(x="species",y="sepal_length",data=df)

plt.title("A Box Plot of Sepal Length by Species")
plt.show()
```

```python
#Creating a Scatter Plot to visualize the relationship between sepal length and sepal width
sns.scatterplot(x="sepal_length",y="sepal_width",hue="species",data=df)
plt.title("Scatter Plot of Sepal Length vs Sepal Width")
plt.show()
```
